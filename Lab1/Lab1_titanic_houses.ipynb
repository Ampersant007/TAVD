```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ds_titanic = pd.read_csv("titanic.csv")

print(f"\nКолонки: {ds_titanic.columns.tolist()}")

plt.figure(figsize=(10,6))
sns.countplot(data=ds_titanic,
             x='Pclass',
             hue='2urvived',
             palette='Set2')
plt.title('Выживаемость пассажиров по классу билета', fontsize=14)
plt.xlabel('Класс билета', fontsize=12)
plt.ylabel('Количество пассажиров', fontsize=12)
plt.legend(['Погиб', 'Выжил'], title='Статус')
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

```python
ds_housing = pd.read_csv("housing.csv")
print(f"\nКолонки: {ds_housing.columns.tolist()}")

plt.figure(figsize=(10,6))
sns.countplot(data=ds_housing,
            x='ocean_proximity')
plt.title('Количество домов по близости к океану', fontsize=14)
plt.xlabel('Близость к океану', fontsize=12)
plt.ylabel('Количество домов', fontsize=12)
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```
